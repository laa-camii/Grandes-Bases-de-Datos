```python
#Cargar la base de datos de Estudiantes.
Estudiantes1 = pd.read_csv("EjemploEstudiantes.csv")
Estudiantes1
```

```python
#Extraccción de etiquetas de los alumnos
etiquetas = Estudiantes1["Alumno"]
etiquetas
```

Aquí guardamos los nombres de los alumnos por separado. Esto nos servirá más adelante para poder identificarlos en las gráficas.

```python
from numpy import array
from sklearn.decomposition import TruncatedSVD
A = Estudiantes1

#Remover columna no numérica.
A = A.drop(["Alumno"], axis=1)
A
```

Aquí eliminamos la columna de Alumno, ya que contiene nombres y para realizar la descomposición necesitamos trabajar únicamente con los datos numéricos.

```python
from numpy import diag
from numpy import zeros
from scipy.linalg import svd
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#SVD
U, s, VT = svd(A) #Obtención de matrices cuyo producto arroja la matriz A
Sigma = zeros((A.shape[0],A.shape[1])) #Creación de matrices de ceros de la misma dimensión de A
Sigma[:A.shape[1], :A.shape[1]] = diag(s)
```

Aquí aplicamos el método SVD a nuestra matriz de calificaciones. Con esto obtenemos las matrices U, Sigma y VT, que utilizaremos posteriormente para reducir las dimensiones de los datos.

```python
#Matriz U
print("Matriz U:")
print(U)
```

```python
#Matriz Sigma
print("Matriz Sigma")
print(Sigma)
```

```python
#Matriz VT
print("Matriz VT")
print(VT)
```

```python
#Para reconstruir la matriz original 
B = U.dot(Sigma.dot(VT))
print(B)
```

Aquí volvemos a multiplicar las matrices U, Sigma y VT para comprobar que podemos reconstruir la matriz original de calificaciones.

```python
#Reducción de la matriz a dos dimensiones
n_elements = 2

UReduced2 = U[:,:n_elements]
print(UReduced2)
```

En esta parte nos quedamos solamente con los dos primeros componentes. La idea es resumir la información de las cinco materias en solo dos dimensiones.

```python
#Reducción de la matriz Sigma a dos dimensiones
SigmaReduced2 = Sigma[:n_elements,:n_elements]
print("Matriz Sigma Reducida")
print(SigmaReduced2)
```

```python
#Reducción de la matriz VT a dos dimensiones
VTReduced2 = VT[:n_elements,:]
print("VT Reducida")
print(VTReduced2)
```

```python
#Matriz transformada a dos dimensiones
AReduced2 = UReduced2.dot(SigmaReduced2.dot(VTReduced2))
print("Matriz transformada")
print(AReduced2)
```

```python
#Coordenadas en dos dimensiones.
T2 = UReduced2.dot(SigmaReduced2)
print("Matriz T (2 dimensiones)")
print(T2)
```

En esta parte obtenemos dos nuevas coordenadas para cada estudiante. Estas coordenadas serán utilizadas para representarlos en una gráfica de dos dimensiones.

```python
#Gráfica de dispersión en dos dimensiones.
df = pd.DataFrame(T2, columns=["a","b"])
df.plot(kind = "scatter", x="a",y="b")
plt.show
```

```python
df = pd.DataFrame(T2, columns=["a","b"])
x = df.iloc[:,0]
y = df.iloc[:,1]
x = x.to_numpy()
y = y.to_numpy()

fig,ax= plt.subplots()
ax.set_title("Mapa de observaciones")
ax.scatter(x,y)

for i, txt in enumerate (etiquetas):
    ax.annotate(txt,(x[i], y[i]))
```

```python
#Reducción a 3 dimensiones
n_elements = 3

UReduced3 = U[:,:n_elements]
SigmaReduced3 = Sigma[:n_elements,:n_elements]
VTReduced3 = VT[:n_elements,:]
AReduced3 = UReduced3.dot(SigmaReduced3.dot(VTReduced3))
print("Matriz transformada")
print(AReduced3)
```

Ahora repetimos el procedimiento, pero conservando tres componentes en lugar de dos. Esto permite conservar más información de la matriz original.

```python
#Coordenadas en tres dimensiones
T3 = UReduced3.dot(SigmaReduced3)
print(T3)
```

```python
from mpl_toolkits import mplot3d
%matplotlib inline
```

```python
df3 = pd.DataFrame(T3, columns = ["a" , "b", "c"])
df3
```

```python
fig = plt.figure()
ax = plt.axes(projection="3d")
xline = df3["b"]
yline = df3["c"]
zline = df3["a"]
ax.scatter3D(xline, yline, zline)

for i, nombre in enumerate(etiquetas):
    ax.text(xline.iloc[i], yline.iloc[i], zline.iloc[i], nombre)
```

Aquí representamos a los estudiantes utilizando tres dimensiones. Esto nos permite observar con más detalle cuáles estudiantes se encuentran cerca y podrían tener comportamientos similares en sus calificaciones.

```python
plt.figure(figsize=(10,7))
Error2 = B-AReduced2

dfError2 = pd.DataFrame(Error2, columns=["Matematicas", "Ciencias", "Espanol", "Historia", "EdFisica"])
dfError2.boxplot()
```

Aquí utilizamos un boxplot para observar cómo se distribuyen los errores de la reconstrucción en cada una de las materias.

```python
print("Errores promedios por columna")
Error2.mean(axis=0)
```

En este bloque calculamos el error promedio de cada materia para observar qué tan cercana quedó la reconstrucción de dos dimensiones a los datos originales.

```python
#Boxplot de los errores con tres dimensiones.
Error3 = B-AReduced3

dfError3= pd.DataFrame(Error3, columns=["Matematicas", "Ciencias", "Espanol", "Historia", "EdFisica"])
dfError3.boxplot()
```

```python
#Error promedio con tres dimensiones
print("Errores promedios por columna")
Error3.mean(axis=0)
```

### ¿Cuál de las separaciones previas es más clara?
En este caso resulta ser más clara la separación en tres dimensiones mediante Single Value Decomposition (SVD), ya que en este ejercicio tenemos pocos estudiantes y la tercera dimensión nos permite ver de mejor forma las diferencias y agrupaciones entre ellos.

Aunque analizar la gráfica en dos dimensiones es más sencillo, con la representación en tres dimensiones conservamos más información y podemos distinguir mejor la posición en la que se encuentra cada estudiante.

En lo personal, creo que en este caso es sencillo interpretar la gráfica en 3D, pero estoy consciente de que probablemente en bases de datos con una mayor cantidad de información esta representación podría volverse un poco más complicada de analizar.

### ¿Difieren sus resultados de los obtenidos en ACP?
Al comparar estos resultados con el Análisis de Componentes Principales (ACP), las posiciones no son exactamente iguales, pero podemos observar de forma muy sencilla agrupaciones similares de estudiantes.

Esto nos indica que ambos métodos nos permiten identificar a los estudiantes que presentan comportamientos similares, aunque la forma de realizar la reducción de dimensiones es diferente.
